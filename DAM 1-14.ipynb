```python
rules = {
    # aircraft has 180 seats
    "Passengers": (0, 180),
    # domestic sector in INR
    "Ticket_Price": (1000, 20000),
    # a delay cannot be negative
    "Delay_Minutes": (0, 600),
    "Flight_Duration_Min": (60, 240),
    # rating scale 1 to 5
    "Customer_Rating": (1, 5)
}

for col, (low, high) in rules.items():
    bad = raw[(raw[col] < low) | (raw[col] > high)]
    print(col, "-> impossible values:", len(bad))
    if len(bad) > 0:
        print(bad[["Date", "Flight_ID", col]].to_string(index=False))
    print()
```

#STEP 4: Removing impossible Values

```python
import numpy as np

# Create a copy
df = raw.copy()

# Find ratings below 1 or above 5
bad_values = (df["Customer_Rating"] < 1) | (df["Customer_Rating"] > 5)

# Count unusual ratings
print("Unusual ratings:", bad_values.sum())

# Replace unusual ratings with missing values
df.loc[bad_values, "Customer_Rating"] = np.nan

# Compare missing cells
print("Missing cells before:", raw.isnull().sum().sum())

print("Missing cells after:", df.isnull().sum().sum())
```

#STEP 5: Capping (handling outliers)

```python

# Select the variable
col = "Ancillary_Revenue"

# Start with the original, uncapped data
before = raw[col].copy()

# Create a fresh dataset for capping
df_capped = raw.copy()

# Calculate Q1, Q3 and IQR using original data
Q1 = before.quantile(0.25)
Q3 = before.quantile(0.75)
IQR = Q3 - Q1

# Calculate the upper and lower limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Apply capping to the new dataset
df_capped[col] = before.clip(lower, upper)

# Print maximum values
print("Maximum before:", before.max())
print("Maximum after:", df_capped[col].max())
print("Upper limit:", upper)

# Plot before and after
plt.figure(figsize=(12, 5))

plt.plot(raw["Date"], before,
         label="Before Capping", marker="o")

plt.plot(raw["Date"], df_capped[col],
         label="After Capping", marker="x")

plt.title("Ancillary Revenue: Before vs After Capping")
plt.xlabel("Date")
plt.ylabel("Ancillary Revenue (₹)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

METHOD 2: IMPUTATION WITH MEAN

```python
# Create a copy of the original dataset
df_mean = df.copy()

# Display descriptive statistics BEFORE filling missing values
print("DESCRIPTIVE STATISTICS BEFORE IMPUTATION")
display(df.describe().round(2))

# Fill missing values in numeric columns using the mean
df_mean[num_cols] = df_mean[num_cols].fillna(df_mean[num_cols].mean())

# Fill missing Route values using the most common route (mode)
df_mean["Route"] = df_mean["Route"].fillna(df_mean["Route"].mode()[0])

# Display descriptive statistics AFTER filling missing values
print("DESCRIPTIVE STATISTICS AFTER MEAN AND MODE IMPUTATION")
display(df_mean.describe().round(2))

# Compare the number of rows before and after
print("Rows before:", len(df))
print("Rows after :", len(df_mean))

#Count missing cells remaining after imputation
print("Blanks left:",df_mean.isnull().sum().sum())
```

METHOD 3: IMPUTATION WITH MODE

```python
df_mode = df.copy()

for col in df_mode.columns:
    df_mode[col] = df_mode[col].fillna(df_mode[col].mode()[0])

    # Display descriptive statistics AFTER filling missing values
print("DESCRIPTIVE STATISTICS AFTER  MODE IMPUTATION")
display(df_mode.describe().round(2))

print("Blanks left:", df_mode.isnull().sum().sum())
```

METHOD 4: IMPUTATION WITH MEDIAN

```python
df_median = df.copy()
df_median[num_cols] = df_median[num_cols].fillna(df_median[num_cols].median())

#Route is text,so use the most common route
df_median["Route"] = df_median["Route"].fillna(df_median["Route"].mode()[0])

# Display descriptive statistics AFTER filling missing values
print("DESCRIPTIVE STATISTICS AFTER MEDIAN IMPUTATION")
display(df_median.describe().round(2))

print("Blanks left:",df_median.isnull().sum().sum())
```

How did the sale build up week by week in each state?-Animation plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import seaborn as sns

df =pd.read_excel("/content/Zepto_Orders_Data (1).xlsx", sheet_name="Orders")
```

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib.pyplot as plt

# Ensure Order_Date is datetime and extract the Week column
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Week'] = df['Order_Date'].dt.isocalendar().week

# Running total sales, week by week for each state
running = df.pivot_table(index="Week", columns="State", values="Order_Value", aggfunc="sum").fillna(0).cumsum()
running = running[running.iloc[-1].sort_values().index]

fig, ax = plt.subplots(figsize=(7, 5))

def draw(i):
    ax.clear()
    # Use the frame index 'i' to display cumulative sales up to that frame's week
    ax.barh(running.columns, running.iloc[i] / 1000, color="#6A189A")
    ax.set_xlim(0, running.values.max() / 1000 * 1.1)
    ax.set_title("Sales up to week " + str(running.index[i]))
    ax.set_xlabel("Sales (INR thousand)")

animation = FuncAnimation(fig, draw, frames=len(running), interval=700)

plt.close()
HTML(animation.to_jshtml())
```

Principal component analysis

```python
groups = {
    "Customer Experience": ["Taste_Score", "Service_Score", "Cleanliness_Score", "Ambience_Score", "Staff_Friendliness_Score"],
    "Digital Engagement": ["App_Orders_Share_Pct", "Loyalty_Members_K", "App_Rating", "Social_Followers_K"],
    "Marketing Intensity": ["Social_Media_Spend_K", "Outdoor_Ad_Spend_K", "Influencer_Spend_K", "Promo_Offers_Count"],
    "Store Capacity": ["Store_Area_SqFt", "Seats", "Staff_Count"],
}

for name, cols in groups.items():
    c = df[cols].corr().values
    average = (c.sum() - len(cols)) / (len(cols) * (len(cols) - 1))  # average correlation inside the group
    print(f"{name}: {len(cols)} variables, average correlation {average:.2f}")
```

```python
df[["Taste_Score", "Service_Score", "Cleanliness_Score", "Ambience_Score", "Staff_Friendliness_Score"]].corr()
```

PCA-1

Customer experience index

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

cols = groups["Customer Experience"]

# Step A: put every variable on the same scale
scaled = StandardScaler().fit_transform(df[cols])

# Step B: run PCA
pca = PCA()
scores = pca.fit_transform(scaled)

# Step C: how much does each component explain?
names = [f"PC{i+1}" for i in range(len(cols))]
print(pd.Series(pca.explained_variance_ratio_, index=names).round(2))
```

```python
# Step D: the loadings show how much each variable counts in PC1

sign = 1 if pca.components_[0].sum() > 0 else -1  # make sure a higher score means better

print(pd.Series(pca.components_[0] * sign, index=cols).round(2))
```

```python
# Step E: turn PC1 into an index from 0 to 100

raw_index = scores[:, 0] * sign

df["Experience_Index"] = (raw_index - raw_index.min()) / (raw_index.max() - raw_index.min()) * 100

variance_explained = {"Experience_Index": round(pca.explained_variance_ratio_[0] * 100)}

print(df[["Store_ID", "Experience_Index"]].sort_values("Experience_Index").head(5).round(1))
df.head()
```

PCA -2

Digital Engagement Index

```python
def make_index(columns, index_name):
    scaled = StandardScaler().fit_transform(df[columns])
    pca = PCA().fit(scaled)
    scores = pca.transform(scaled)[:, 0]

    sign = 1 if pca.components_[0].sum() > 0 else -1
    scores = scores * sign

    df[index_name] = (scores - scores.min()) / (scores.max() - scores.min()) * 100
    variance_explained[index_name] = round(pca.explained_variance_ratio_[0] * 100)

    print("Variance explained by PC1:", variance_explained[index_name], "%")
    print(pd.Series(pca.components_[0] * sign, index=columns).round(2))


make_index(groups["Digital Engagement"], "Digital_Index")
```

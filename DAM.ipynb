```python
# This is a simple calculator that evaluates a string equation.
# WARNING: Using `eval()` can be dangerous if the input is not trusted.
# It allows arbitrary code execution.

while True:
    equation = input("Enter an equation (e.g., '2 + 3 * 4' or '10 / 2 - 1'): ")
    try:
        result = eval(equation)
        print(f"The result of '{equation}' is: {result}")
    except (SyntaxError, NameError, TypeError, ZeroDivisionError) as e:
        print(f"Invalid input or calculation error: {e}")

    another_calculation = input("Perform another calculation? (yes/no): ")
    if another_calculation.lower() != "yes":
        break
```
